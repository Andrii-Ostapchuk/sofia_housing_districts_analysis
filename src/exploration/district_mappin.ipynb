```python
import pandas as pd
import geopandas as gpd

# Gets the folder directory where urban data is located to avoid duplication
folder = r'D:\Data Analysis\5_Projects\sofia_housing_districts_analysis\data\sofia_data\urbandata_sofia_bg'

# Reads file with district boundaries
district_boundaries = gpd.read_file(folder + r'\natovarv_sasht_polojenie_kk_072020_26_sofpl_20201118.geojson')

# Gets unique districts
district_boundaries['rajon'].unique()

districts = district_boundaries['rajon'].unique()

split_districts = set()
for district in districts:
    if isinstance(district, str):
        split_district = district.split(' / ')
        split_districts.add(split_district[0].strip())

    if len(split_district) == 2:
        split_districts.add(split_district[1].strip())

# A set of unique Sofia Municipality districts in cyrillic
split_districts
```
